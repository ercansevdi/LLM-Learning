```python
from sentence_transformers import SentenceTransformer
from sentence_transformers.util import cos_sim

MODEL_NAME = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"

model = SentenceTransformer(MODEL_NAME)

print("Model yüklendi:")
print(MODEL_NAME)
```

```python
sentence = "Elektrik motorunda yüksek titreşim var."

embedding = model.encode(sentence)

print(embedding)
```

```python
print(type(embedding))
print(embedding.shape)
```

```python
sentence1 = "Elektrik motorunda yüksek titreşim var."

sentence2 = "Motor normalden fazla vibrasyon yapıyor."

embedding1 = model.encode(sentence1)

embedding2 = model.encode(sentence2)
```

```python
similarity = cos_sim(
    embedding1,
    embedding2
)

print(similarity)
```

```python
score = similarity.item()

print(
    "Benzerlik skoru:",
    round(score, 4)
)
```

```python
query = "Motorda titreşim problemi var."
```

```python
documents = [
    "Rulman arızası motor titreşiminde artışa neden olabilir.",
    "Motor sıcaklığının yükselmesi aşırı yükten kaynaklanabilir.",
    "PLC giriş modülleri dijital sensörlerden sinyal alır.",
    "Yağlama eksikliği rulman ömrünü azaltabilir.",
    "Bugün fabrikada bakım toplantısı yapılacaktır."
]
```

```python
query_embedding = model.encode(query)
```

```python
document_embeddings = model.encode(
    documents
)
```

```python
print(
    document_embeddings.shape
)
```

```python
scores = cos_sim(
    query_embedding,
    document_embeddings
)

print(scores)
```

```python
for document, score in zip(
    documents,
    scores[0]
):

    print(
        round(score.item(), 4),
        "→",
        document
    )
```

```python
best_index = scores[0].argmax().item()

best_document = documents[
    best_index
]

best_score = scores[0][
    best_index
].item()

print(
    "En ilgili doküman:"
)

print(best_document)

print(
    "\nBenzerlik:",
    round(best_score, 4)
)
```

```python
def semantic_search(
    query,
    documents,
    model
):

    query_embedding = model.encode(
        query
    )

    document_embeddings = model.encode(
        documents
    )

    scores = cos_sim(
        query_embedding,
        document_embeddings
    )[0]

    results = []

    for index, score in enumerate(scores):

        results.append(
            {
                "document": documents[index],
                "score": score.item()
            }
        )

    results = sorted(
        results,
        key=lambda x: x["score"],
        reverse=True
    )

    return results
```

```python
results = semantic_search(
    "Motorda titreşim problemi var.",
    documents,
    model
)
```

```python
for result in results:

    print(
        round(result["score"], 4),
        "→",
        result["document"]
    )
```

```python
query = "Rulman yeterince yağlanmazsa ne olur?"

results = semantic_search(
    query,
    documents,
    model
)

for result in results:

    print(
        round(result["score"], 4),
        "→",
        result["document"]
    )
```
