```python
from transformers import AutoTokenizer

MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Tokenizer başarıyla yüklendi.")
print("Model:", MODEL_NAME)
print("Vocabulary boyutu:", len(tokenizer))
```

```python
from transformers import AutoTokenizer
```

```python
AutoTokenizer.from_pretrained("Qwen/Qwen2.5-0.5B-Instruct")
```

```python
text = "Elektrik motorlarında rulman arızası oluştu."

print(text)
```

```python
tokens = tokenizer.tokenize(text)

print("Tokenlar:")
print(tokens)

print("\nToken sayısı:", len(tokens))
```

```python
token_ids = tokenizer.encode(
    text,
    add_special_tokens=False
)

print(token_ids)
```

```python
for i, (token, token_id) in enumerate(zip(tokens, token_ids), start=1):
    print(f"{i:2} | {token_id:6} | {repr(token)}")
```

```python
decoded_text = tokenizer.decode(
    token_ids,
    skip_special_tokens=True
)

print("Orijinal:")
print(text)

print("\nDecode edilmiş:")
print(decoded_text)
```

```python
test_words = [
    "motor",
    "motorlar",
    "motorlardan",
    "motorlarımızdan",
    "motorlarımızdakilerden"
]

for word in test_words:
    tokens = tokenizer.tokenize(word)

    print(f"\nKelime: {word}")
    print("Tokenlar:", tokens)
    print("Token sayısı:", len(tokens))
```

```python
texts = [
    "Elektrik motorlarında rulman arızası oluştu.",
    "A bearing failure occurred in the electric motors."
]

for text in texts:
    ids = tokenizer.encode(text, add_special_tokens=False)

    print("\nMetin:")
    print(text)

    print("Kelime sayısı:", len(text.split()))
    print("Token sayısı:", len(ids))
    print("Tokenlar:", tokenizer.tokenize(text))
```

```python
def analyze_text(text):

    tokens = tokenizer.tokenize(text)
    token_ids = tokenizer.encode(
        text,
        add_special_tokens=False
    )

    print("=" * 60)
    print("METİN:")
    print(text)

    print("\nKelime sayısı:", len(text.split()))
    print("Token sayısı:", len(token_ids))

    print("\nTOKEN ANALİZİ")
    print("-" * 60)

    for i, (token, token_id) in enumerate(
        zip(tokens, token_ids),
        start=1
    ):
        print(
            f"{i:2} | ID: {token_id:6} | Token: {repr(token)}"
        )

    print("\nDECODE:")
    print(
        tokenizer.decode(
            token_ids,
            skip_special_tokens=True
        )
    )
```

```python
analyze_text(
    "STM32 mikrodenetleyicilerinde TinyML modeli çalıştırıyorum."
)
```

```python
analyze_text(
    "Akıllı fabrikalarda otonom taşıma araçları kullanılmaktadır."
)
```
